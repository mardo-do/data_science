```python
# Import all required libraries
import yfinance as yf
import pandas as pd
import requests
from bs4 import BeautifulSoup
import matplotlib.pyplot as plt
```

```python
print("Exercise 1: Extracting Tesla Stock Data")
print("=" * 50)

# 1. Use yfinance to extract Tesla stock data
tesla = yf.Ticker("TSLA")

# 2. Get historical price data
tesla_data = tesla.history(period="max")

# 3. Reset the index
tesla_data.reset_index(inplace=True)

# 4. Display the first five rows
print("First five rows of tesla_data:")
print(tesla_data.head())
print("\n" + "=" * 50)
```

```python
print("Exercise 2: Extracting Tesla Revenue Data")
print("=" * 50)

# 1. Define the URL for Tesla revenue data
url = "https://www.macrotrends.net/stocks/charts/TSLA/tesla/revenue"

# 2. Download the webpage
headers = {"User-Agent": "Mozilla/5.0"}
html_data = requests.get(url, headers=headers).text

# 3. Parse the HTML
soup = BeautifulSoup(html_data, "html5lib")

# 4. Find all tables
tables = soup.find_all("table")

# 5. Identify and extract the revenue table (usually the second table)
# Let's find the table containing "Quarterly Revenue"
for i, table in enumerate(tables):
    if "Quarterly Revenue" in str(table):
        revenue_table_index = i
        break
else:
    revenue_table_index = 1  # Default to second table if not found

# 6. Read the table into DataFrame
tesla_revenue = pd.read_html(str(tables[revenue_table_index]))[0]

# 7. Rename columns
tesla_revenue.columns = ["Date", "Revenue"]

# 8. Clean the Revenue column: remove commas, dollar signs, and empty values
tesla_revenue["Revenue"] = tesla_revenue["Revenue"].str.replace(",", "").str.replace("$", "").str.replace(" ", "")

# 9. Remove rows with empty revenue
tesla_revenue = tesla_revenue[tesla_revenue["Revenue"] != ""]

# 10. Display the last five rows
print("Last five rows of tesla_revenue:")
print(tesla_revenue.tail())
print("\n" + "=" * 50)
```

```python
print("Exercise 3: Extracting GameStop Stock Data")
print("=" * 50)

# 1. Use yfinance to extract GameStop stock data
gme = yf.Ticker("GME")

# 2. Get historical price data
gme_data = gme.history(period="max")

# 3. Reset the index
gme_data.reset_index(inplace=True)

# 4. Display the first five rows
print("First five rows of gme_data:")
print(gme_data.head())
print("\n" + "=" * 50)
```

```python
print("Exercise 4: Extracting GameStop Revenue Data")
print("=" * 50)

# 1. Define the URL for GameStop revenue data
url_gme = "https://www.macrotrends.net/stocks/charts/GME/gamestop/revenue"

# 2. Download the webpage
html_data_gme = requests.get(url_gme, headers=headers).text

# 3. Parse the HTML
soup_gme = BeautifulSoup(html_data_gme, "html5lib")

# 4. Find all tables
tables_gme = soup_gme.find_all("table")

# 5. Identify and extract the revenue table
for i, table in enumerate(tables_gme):
    if "Quarterly Revenue" in str(table):
        revenue_table_index_gme = i
        break
else:
    revenue_table_index_gme = 1

# 6. Read the table into DataFrame
gme_revenue = pd.read_html(str(tables_gme[revenue_table_index_gme]))[0]

# 7. Rename columns
gme_revenue.columns = ["Date", "Revenue"]

# 8. Clean the Revenue column
gme_revenue["Revenue"] = gme_revenue["Revenue"].str.replace(",", "").str.replace("$", "").str.replace(" ", "")

# 9. Remove rows with empty revenue
gme_revenue = gme_revenue[gme_revenue["Revenue"] != ""]

# 10. Display the last five rows
print("Last five rows of gme_revenue:")
print(gme_revenue.tail())
print("\n" + "=" * 50)
```

```python
print("Exercise 5: Plotting Tesla Stock Price and Revenue")
print("=" * 50)

# 1. Prepare data for plotting
# Convert Revenue to numeric
tesla_revenue["Revenue"] = pd.to_numeric(tesla_revenue["Revenue"])

# Convert Date columns to datetime
tesla_data["Date"] = pd.to_datetime(tesla_data["Date"])
tesla_revenue["Date"] = pd.to_datetime(tesla_revenue["Date"])

# 2. Create the figure with two y-axes
fig, ax1 = plt.subplots(figsize=(14, 8))

# 3. Plot Tesla Stock Price (left y-axis)
color = 'tab:red'
ax1.set_xlabel('Date', fontsize=14)
ax1.set_ylabel('Tesla Stock Price (USD)', color=color, fontsize=14)
ax1.plot(tesla_data["Date"], tesla_data["Close"], color=color, linewidth=2)
ax1.tick_params(axis='y', labelcolor=color)
ax1.grid(True, alpha=0.3)

# 4. Create a second y-axis for Revenue
ax2 = ax1.twinx()
color = 'tab:blue'
ax2.set_ylabel('Tesla Revenue (USD Millions)', color=color, fontsize=14)
ax2.plot(tesla_revenue["Date"], tesla_revenue["Revenue"], color=color, linewidth=2, linestyle='--')
ax2.tick_params(axis='y', labelcolor=color)

# 5. Add title and layout
plt.title('Tesla: Stock Price vs Revenue (2010-2023)', fontsize=16, fontweight='bold')
fig.tight_layout()

print("Displaying Tesla stock price and revenue graph...")
plt.show()

# 6. Display both graphs separately as well
print("\nTesla Stock Price Over Time:")
fig1, ax1 = plt.subplots(figsize=(12, 6))
ax1.plot(tesla_data["Date"], tesla_data["Close"], color='green', linewidth=2)
ax1.set_xlabel('Date')
ax1.set_ylabel('Stock Price (USD)')
ax1.set_title('Tesla Historical Stock Price')
ax1.grid(True, alpha=0.3)
plt.show()

print("\nTesla Revenue Over Time:")
fig2, ax2 = plt.subplots(figsize=(12, 6))
ax2.plot(tesla_revenue["Date"], tesla_revenue["Revenue"], color='purple', linewidth=2)
ax2.set_xlabel('Date')
ax2.set_ylabel('Revenue (USD Millions)')
ax2.set_title('Tesla Historical Revenue')
ax2.grid(True, alpha=0.3)
plt.show()

print("\n" + "=" * 50)
```
